### 🍕 Customer Churn Proportion (Pie Chart)

````python
plt.figure(figsize=(8, 8))
colors = ['#66b3ff', '#ff9999']
explode = (0, 0.1) 
df["Attrition_Flag"].value_counts().plot(
    kind="pie", 
    autopct="%1.1f%%", 
    startangle=90, 
    colors=colors, 
    explode=explode,
    shadow=True,
    textprops={'fontsize': 12, 'weight': 'bold'}
)
plt.title("Proportion of Existing vs. Attrited Customers", fontsize=14, fontweight='bold')
plt.ylabel("")  
plt.show()
````

**Output Analysis:** This code creates a 3D-styled pie chart to show the percentage of active customers versus those who have churned (`Attrition_Flag`). The `explode` feature pulls out the churned slice slightly, making it instantly clear how big or small the customer loss is relative to the total base.

**Output Analysis:** This code generates a scatter plot with a fitted regression line (`sns.regplot`) to analyze the relationship between a customer's age and their tenure (`Months_on_book`). ### 📊 Key Observations from Exploratory Data Analysis (EDA)

#### 1. Data Distribution (Histogram & Boxplot Insights)
* **Customer Age:** The age of customers follows a normal distribution (bell-shaped curve). Most of the bank's customers are between **40 to 50 years old**.
* **Credit Limit:** The credit limit data is heavily skewed to the right. A large majority of customers have a lower credit limit (around $2,000 to $5,000), while only a few premium customers have a very high limit ($34,500).
* **Outliers:** Columns like `Credit_Limit` and `Total_Trans_Amt` contain several outlier points, representing high-net-worth individuals or extreme spenders.

---

#### 2. Categorical Insights (Value Counts & Countplots)
* **Customer Status:** Around **16% of customers have left the bank (Attrited Customers)**, while the remaining 84% are still active (Existing Customers).
* **Card Type:** The **Blue Card** is the most popular card category with the highest number of users. Premium cards like Silver, Gold, and Platinum have very few users.
* **Income Groups:** The largest customer segment belongs to the lower income category (Less than $40K annual salary).

---

#### 3. Relationships & Trends (Scatter Plot & Regression Insights)
* **Age vs. Tenure:** There is a strong positive relationship between `Customer_Age` and `Months_on_book`. Older customers have generally been with the bank for a much longer time.
* **Transactions vs. Spending:** The scatter plot between `Total_Trans_Ct` (number of times card used) and `Total_Trans_Amt` (money spent) shows clear clusters. Customers who use their cards frequently also tend to spend more money overall.
* **Credit Limit vs. Card Utilization:** There is a negative correlation between credit limit and card utilization ratio. Customers with very high credit limits generally utilize a smaller percentage of their available credit limit.

---

#### 4. Group Comparisons (Groupby & Bar/Boxplot Insights)
* **Income vs. Credit Limit:** On average, customers with higher annual income categories are granted higher credit limits by the bank.
* **Gender-Based Trends:** Male customers, on average, hold higher credit limits compared to female customers within the same income categories.
* **Churn Analysis:** Customers who leave the bank (Attrited) generally show a significantly lower number of total transaction counts and lower total transaction amounts compared to active customers.

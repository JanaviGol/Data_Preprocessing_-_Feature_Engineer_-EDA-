## 🧮 Phase 2: Advanced Data Imputation Engine

### 📊 5. Single-Variable Baseline Mean Imputation
* **Technical Definition:** A univariate restoration method that calculates the central arithmetic average of an active continuous attribute and maps that constant into missing fields.
* **Project Application:** Successfully resolved missing fields within the continuous `bmi` feature using `SimpleImputer(strategy='mean')`, driving its null tracking metrics to absolute zero.

### 🏷️ 6. Categorical Most Frequent (Mode) Imputation
* **Technical Definition:** A discrete attribute handling strategy that fills null locations using the statistical mode, representing the highest frequency value within that column.
* **Project Application:** Simultaneously repaired missing entries within the string-based `gender` and `region` features using `SimpleImputer(strategy='most_frequent')` without changing class labels.

### 🎲 7. Missing Tracking Indicators & Random Sample Allocation
* **Technical Definition:** A dual-layer imputation technique that generates a binary column to flag missing rows while dynamically replacing blanks with randomized observations from the clean subset.
* **Project Application:** Created the tracking column `age_isna` and imputed missing `age` entries using random samples, preserving the original dataset's historical distribution curve perfectly.

### 🤖 8. K-Nearest Neighbors (KNN) Multivariate Imputation
* **Technical Definition:** An advanced spatial distance algorithm that uses Euclidean distance models ($k$-neighbors) to compute localized predictions for missing continuous data points.
* **Project Application:** Replaced basic statistical averages by mapping data relationships in `cholesterol` with a 3-neighbor spatial matrix using `KNNImputer(n_neighbors=3)`.

### ⛓️ 9. Multivariate Imputation by Chained Equations (MICE)
* **Technical Definition:** A highly sophisticated iterative algorithm that treats every missing column as a target variable in a series of continuous linear regression predictions.
* **Project Application:** Deployed via `IterativeImputer` to run 10 optimization cycles over `glucose`, completing the missing data handling phase with **0% data loss**.

## 📏 Phase 3: Statistical Outlier Engineering

### ✂️ 10. Normal Distribution Z-Score Trimming
* **Technical Definition:** An outlier filtering technique that computes standard deviation distances from the mean, dropping data points that cross a set threshold ($|Z| > 3$).
* **Project Application:** Scanned for extreme anomalies in `cholesterol` and `glucose`, resulting in the hard row-deletion of 32 combined extreme rows to stabilize the overall variance.

### 📦 11. Non-Parametric Interquartile Range (IQR) Trimming
* **Technical Definition:** A distribution-free outlier filtering framework that establishes hard boundaries based on quartiles, using the mathematical formula: $[Q1 - 1.5 \times IQR, Q3 + 1.5 \times IQR]$.
* **Project Application:** Successfully located 19 out-of-boundary patient records inside the skewed `bmi` distribution, calculating exact lower and upper fences for testing purposes.

### 📌 12. Fixed Percentile Boundary Trimming & Clipping
* **Technical Definition:** An outlier correction method that forces extreme data points outside a set range back within strict, pre-defined percentile limits using `np.clip`.
* **Project Application:** Enforced strict boundaries on the `bmi` feature by clipping values at the 1st and 99th percentiles, reducing extreme values without deleting any records.

### 🪵 13. Symmetrical Multivariate Outlier Winsorization
* **Technical Definition:** A robust statistical transform that clips extreme distribution tails symmetrically from both sides at a designated percentage, replacing outliers with maximum safe values.
* **Project Application:** Iterated a 5% symmetric `winsorize` filter over all continuous profiles, neutralizing anomalies while keeping all **400 rows** completely active for machine learning modeling.

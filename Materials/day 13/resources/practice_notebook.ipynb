### Feature Construction


```python
import pandas as pd

# Sample real estate dataset
df = pd.DataFrame({
    'price':     [5000000, 8000000, 3500000, 12000000],
    'area_sqft': [850,      1200,    650,     1800],
    'bedrooms':  [2,         3,       1,        4],
    'date_listed': pd.to_datetime(['2024-01-15', '2024-03-22',
                                   '2024-02-08', '2024-04-01'])
})

# ✅ Arithmetic Feature
df['price_per_sqft'] = df['price'] / df['area_sqft']

# ✅ Ratio Feature
df['price_per_bed'] = df['price'] / df['bedrooms']

# ✅ Datetime Feature
df['listing_month']   = df['date_listed'].dt.month
df['listing_weekday'] = df['date_listed'].dt.day_name()

# ✅ Polynomial/Interaction Feature
df['area_squared'] = df['area_sqft'] ** 2


df[['price', 'price_per_sqft', 'price_per_bed', 'listing_month']]
```

### Feature Splitting


```python
import pandas as pd

df = pd.DataFrame({
    'full_name':    ['Rahul Sharma', 'Priya Patel', 'Arjun Singh'],
    'location':     ['Mumbai, Maharashtra', 'Ahmedabad, Gujarat', 'Delhi, Delhi'],
    'order_date':   ['2024-03-15 14:30:00', '2024-03-16 09:00:00', '2024-03-17 18:45:00'],
    'product_code': ['ELEC-1042', 'CLOTH-2031', 'FOOD-0098']
})

# ✅ Split full_name into first and last name
df[['first_name', 'last_name']] = df['full_name'].str.split(' ', n=1, expand=True)

# ✅ Split location into city and state
df[['city', 'state']] = df['location'].str.split(', ', expand=True)

# ✅ Split datetime into date and time
df['order_dt'] = pd.to_datetime(df['order_date'])
df['date_only'] = df['order_dt'].dt.date
df['hour']      = df['order_dt'].dt.hour

# ✅ Split product_code into category and prod_id
df[['category', 'prod_id']] = df['product_code'].str.split('-', expand=True)

df[['first_name', 'city', 'state', 'hour', 'category', 'prod_id']]
```

### FunctionTransformer


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import FunctionTransformer

# Sample skewed salary data (in lakhs) - this type of data often benefits from transformation
df = pd.DataFrame({'Salary_Lakhs': [2, 5, 8, 15, 25, 50, 120, 300]})

print("Original Salary Data (Lakhs):")
print(df['Salary_Lakhs'].values)

# === Usage 2: Applying a simpler custom mathematical transformation (Squaring) ===
# FunctionTransformer can wrap any Python function, allowing you to integrate custom logic.
# Here, we'll demonstrate a simple squaring function.
def square_func(X):
    return X ** 2

# Create a transformer for our custom function
square_transformer = FunctionTransformer(func=square_func)

# Apply the custom transformation
df['Salary_Squared'] = square_transformer.fit_transform(df[['Salary_Lakhs']])

print("\nSalaries after Squaring Transformation:")
print(df['Salary_Squared'].values.round(3))

print("\nFunctionTransformer is a versatile tool to integrate any transformation into your scikit-learn workflow!")

df[['Salary_Lakhs', 'Salary_Squared']]
```

### Log Transform


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import FunctionTransformer

# Real India example: House prices in Mumbai (lakhs)
df = pd.DataFrame({
    'price_lakh': [35, 42, 55, 80, 120, 200, 350, 800, 1500, 5000]
})

print("SKEWNESS COMPARISON")
print(f"Original skewness        : {df['price_lakh'].skew():.3f}")

# ✅ Log Transform using FunctionTransformer
log_transformer = FunctionTransformer(
    func=np.log1p,
    inverse_func=np.expm1,
    validate=True
)

df['price_log'] = log_transformer.fit_transform(df[['price_lakh']])

print(f"Log transformed skewness : {df['price_log'].skew():.3f}")
print()
df[['price_lakh', 'price_log']]
```

### Reciprocal Transform


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import FunctionTransformer

# 🏍️ Delivery data
data = pd.DataFrame({
    'vehicle': ['Cycle', 'Scooter', 'Bike', 'Car', 'Bullet Train'],
    'speed_kmph': [10, 40, 60, 100, 300]
})

# # 🔄 Reciprocal Transform (safe version with +1)
recip_transformer = FunctionTransformer(func=lambda x: 1 / (x + 1))

data['speed_reciprocal'] = recip_transformer.transform(data[['speed_kmph']])

# Also show simple 1/x for non-zero values
data['speed_1_over_x'] = 1 / (data['speed_kmph'] + 1)

data
```

### Square Root Transform


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import FunctionTransformer

# 📱 Social media engagement data
posts = pd.DataFrame({
    'post_type': ['Text', 'Image', 'Reel', 'Story', 'Viral Reel'],
    'likes': [15, 150, 2500, 45, 50000],
    'comments': [3, 25, 400, 8, 8000]
})

# 🌿 Square Root Transform
sqrt_transformer = FunctionTransformer(
    func=np.sqrt,
    inverse_func=np.square
)

posts['likes_sqrt'] = sqrt_transformer.transform(posts[['likes']])
posts['comments_sqrt'] = sqrt_transformer.transform(posts[['comments']])

# 🔄 Verify inverse
posts['likes_restored'] = sqrt_transformer.inverse_transform(posts[['likes_sqrt']]).round(0)
posts['comments_restored'] = sqrt_transformer.inverse_transform(posts[['comments_sqrt']]).round(0)

posts[['post_type','likes','likes_sqrt','likes_restored','comments','comments_sqrt', 'comments_restored']]
```

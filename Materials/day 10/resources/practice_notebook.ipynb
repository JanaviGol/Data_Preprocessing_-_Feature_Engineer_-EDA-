### Ordinal Encoding


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OrdinalEncoder

# ─── Dataset: HDFC Loan Applications ───
data = {
    'applicant': ['Aarav', 'Priya', 'Ravi', 'Sneha', 'Karan',
                  'Divya', 'Mohit', 'Ananya'],
    'education': ['Graduate',      'Post-Graduate', 'School',
                  'Graduate',      'Post-Graduate', 'School',
                  'Graduate',      'Post-Graduate'],
    'income_level': ['Low',    'High',   'Medium', 'High',
                     'Low',    'Medium', 'High',   'Low'],
    'loan_approved': [1, 1, 0, 1, 0, 1, 1, 0],
}
df = pd.DataFrame(data)

print("Original Data:")
df
```

```python
# ════════════════════════════════════════
# METHOD 1: Manual Map — Sabse Simple, Full Control
# ════════════════════════════════════════
education_order = {'School': 1, 'Graduate': 2, 'Post-Graduate': 3}
income_order    = {'Low': 1, 'Medium': 2, 'High': 3}

df['education_enc']    = df['education'].map(education_order)
df['income_level_enc'] = df['income_level'].map(income_order)

print("\n--- Method 1: Manual Map ---")
df[['education', 'education_enc', 'income_level', 'income_level_enc']]
```

```python
# ════════════════════════════════════════
# METHOD 2: sklearn OrdinalEncoder
# (Machine Learning pipelines ke liye best)
# ════════════════════════════════════════
enc = OrdinalEncoder(
    categories=[
        ['School', 'Graduate', 'Post-Graduate'],   # education order
        ['Low', 'Medium', 'High'],                  # income order
    ]
)

df_encoded = df[['education', 'income_level']].copy()
df[['edu_sklearn', 'income_sklearn']] = enc.fit_transform(df_encoded)

print("\n--- Method 2: sklearn OrdinalEncoder ---")
df[['education', 'edu_sklearn', 'income_level', 'income_sklearn']]
# Note: sklearn 0-indexed (School=0, Graduate=1, Post-Graduate=2)
# Manual map 1-indexed (School=1, Graduate=2, Post-Graduate=3)
```

### Label Encoding


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# ─── Dataset: Ola Driver Classification ───
data = {
    'driver_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'city':      ['Mumbai', 'Delhi', 'Pune', 'Mumbai',
                  'Chennai', 'Delhi', 'Pune', 'Chennai'],
    'vehicle':   ['Auto', 'Bike', 'Car', 'Car',
                  'Auto', 'Car', 'Bike', 'Auto'],
    'gender':    ['Male', 'Female', 'Male', 'Female',
                  'Male', 'Male', 'Female', 'Female'],
    'churned':   ['Yes', 'No', 'No', 'Yes',
                  'No', 'Yes', 'No', 'No'],   # Target variable!
}
df = pd.DataFrame(data)

print("Original Data:")
df
```

```python
# ════════════════════════════════════════
# METHOD 1: sklearn LabelEncoder — TARGET ke liye best
# ════════════════════════════════════════
le = LabelEncoder()

# Target variable encode karo
df['churned_enc'] = le.fit_transform(df['churned'])
print("\n--- Target Variable (churned) Encoded ---")
print(f"Classes: {le.classes_}")   # ['No', 'Yes']
print(f"Mapping: No→{le.transform(['No'])[0]}, Yes→{le.transform(['Yes'])[0]}")
df[['churned', 'churned_enc']]
```

```python
# ════════════════════════════════════════
# METHOD 2: Alag alag columns encode karo
# ════════════════════════════════════════
le_city    = LabelEncoder()
le_vehicle = LabelEncoder()
le_gender  = LabelEncoder()

df['city_enc']    = le_city.fit_transform(df['city'])
df['vehicle_enc'] = le_vehicle.fit_transform(df['vehicle'])
df['gender_enc']  = le_gender.fit_transform(df['gender'])

print("\n--- Feature Columns Encoded ---")
print(df[['city', 'city_enc', 'vehicle', 'vehicle_enc', 'gender', 'gender_enc']])

# Mappings dekho
print("\nCity Mapping:")
for label, num in zip(le_city.classes_, range(len(le_city.classes_))):
    print(f"  {label} → {num}")

# ════════════════════════════════════════
# Inverse transform — numbers wapas strings mein
# ════════════════════════════════════════
print("\nInverse Transform example:")
print(le.inverse_transform([0, 1, 0]))  # [0,1,0] → ['No','Yes','No']
```

### One Hot Encoding

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder

# ─── Dataset: Swiggy Restaurant Orders ───
data = {
    'order_id':  [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    'city':      ['Mumbai',    'Delhi',    'Pune',     'Mumbai',
                  'Chennai',   'Delhi',    'Pune',     'Chennai'],
    'cuisine':   ['Indian',    'Chinese',  'Italian',  'Indian',
                  'Chinese',   'Italian',  'Indian',   'Chinese'],
    'payment':   ['UPI',       'Card',     'Cash',     'UPI',
                  'Card',      'UPI',      'Cash',     'Card'],
    'amount':    [350, 520, 180, 430, 290, 610, 240, 380],
}
df = pd.DataFrame(data)

print(f"\nOriginal shape: {df.shape}")
print("Original Data:")
df
```

```python
# ════════════════════════════════════════
# METHOD 1: pandas get_dummies — Quick & Easy
# ════════════════════════════════════════
# Ek column ke liye
df_city_ohe = pd.get_dummies(df['city'], prefix='city')
print("\n--- City OHE (get_dummies) ---")
print(df_city_ohe)

# Sabhi categorical columns ek saath
df_ohe = pd.get_dummies(
    df,
    columns=['city', 'cuisine', 'payment'],
    prefix=['city', 'cuisine', 'pay'],
    drop_first=True     # Dummy variable trap se bacho!
)

print("\n--- Full OHE (all categorical columns) ---")
print(df_ohe.to_string())
print(f"New shape: {df_ohe.shape}")  # zyada columns!
print(f"New columns: {list(df_ohe.columns)}")
```

```python
# ════════════════════════════════════════
# METHOD 2: sklearn OneHotEncoder
# (Pipeline mein use ke liye best)
# ════════════════════════════════════════
ohe = OneHotEncoder(
    drop='first',           # Dummy trap se bacho
    sparse_output=False,    # Dense array chahiye
    handle_unknown='ignore' # Test mein naya value aaye toh ignore
)

# Sirf categorical columns
cat_cols = ['city', 'cuisine', 'payment']
encoded_arr = ohe.fit_transform(df[cat_cols])

# Feature names nikalte hain
feature_names = ohe.get_feature_names_out(cat_cols)

df_sklearn_ohe = pd.DataFrame(encoded_arr, columns=feature_names)

print("\n--- sklearn OneHotEncoder ---")
print(f"\nGenerated feature names: {feature_names.tolist()}")
df_sklearn_ohe
```
